# Step 1: Normalise the data for better conditioning

## Normalise data & divide into training, testing, and validation sets

```python
import pandas as pd
df = pd.read_csv("/Users/Tanisha/Desktop/career/2026-27 career/Machine learning project/Machine-learning/Real estate.csv")
train_df = df.sample(frac=0.8, random_state=42).drop(columns = ["No"])
remaining_df = df.drop(train_df.index)
val_df = remaining_df.sample(frac=0.5, random_state=42) #rest for validation + testing
test_df = remaining_df.drop(val_df.index)
train_df_s = (train_df - train_df.mean()) / train_df.std() #normalise data
#randomly select 80% for training
```

## Hessian Spectrum and Condition Number, ${\kappa}$, Analysis: Quantifying Landscape Geometry Post-Normalization

```python
import numpy as np
Z = train_df_s.drop(columns = ["Y house price of unit area"]).to_numpy() #331 x 6 (cause 2 cols dropped, response and 'No')
Hessian = (Z.T @ Z)* 2/331 #6 x 6
eigenvals_array = np.linalg.eigvals(Hessian)
eig_max = max(eigenvals_array)
eig_min = min(eigenvals_array)
condition_no = eig_max/eig_min
print("The condition number post normalisation is", condition_no)
```

```python
Z1 = train_df.drop(columns = ["Y house price of unit area"]).to_numpy()
Hessian_1 = (Z1.T @ Z1)*(2/331)
eigenvals_array_1 = np.linalg.eigvals(Hessian_1)
eig_max_1 = max(eigenvals_array_1)
eig_min_1 = min(eigenvals_array_1)
condition_no_1 = eig_max_1/eig_min_1
print("The condition number pre normalisation is", condition_no_1)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Separate target and extract feature matrices
target_col = 'Y house price of unit area'

# Ensure 'No' is already dropped from your dataframes
Z_raw = train_df.drop(columns=[target_col]).to_numpy()
Z_norm = train_df_s.drop(columns=[target_col]).to_numpy()

N = Z_raw.shape[0]

# 2. Compute Hessians (2/N Z^T Z, or just Z^T Z / N for spectral analysis)
Hessian_raw = (Z_raw.T @ Z_raw) / N
Hessian_norm = (Z_norm.T @ Z_norm) / N

# Use eigvalsh since Z^T Z is symmetric positive semi-definite
eig_raw = np.sort(np.linalg.eigvalsh(Hessian_raw))[::-1]
eig_norm = np.sort(np.linalg.eigvalsh(Hessian_norm))[::-1]

# Calculate condition numbers
kappa_raw = eig_raw[0] / eig_raw[-1]
kappa_norm = eig_norm[0] / eig_norm[-1]

# 3. Plotting the Spectrum Side-by-Side on a Log Scale
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
x_indices = np.arange(len(eig_raw))
feature_labels = [f"$\\lambda_{{{i+1}}}" for i in range(len(eig_raw))]

# Raw Data Spectrum
ax1.bar(x_indices, eig_raw, color='#e74c3c', alpha=0.85, edgecolor='#c0392b')
ax1.set_yscale('log')
ax1.set_title(f"Raw Data Hessian Spectrum\nCondition Number $\\kappa \\approx {kappa_raw:,.1e}$", fontsize=11, fontweight='bold')
ax1.set_ylabel("Eigenvalue Magnitude (Log Scale)", fontsize=10)
ax1.set_xlabel("Eigenvalue Index", fontsize=10)
ax1.set_xticks(x_indices)
ax1.grid(True, which="both", linestyle="--", alpha=0.5)

# Normalized Data Spectrum
ax2.bar(x_indices, eig_norm, color='#2ecc71', alpha=0.85, edgecolor='#27ae60')
ax2.set_yscale('log')
ax2.set_title(f"Normalized Data Hessian Spectrum\nCondition Number $\\kappa \\approx {kappa_norm:.2f}$", fontsize=11, fontweight='bold')
ax2.set_ylabel("Eigenvalue Magnitude (Log Scale)", fontsize=10)
ax2.set_xlabel("Eigenvalue Index", fontsize=10)
ax2.set_xticks(x_indices)
ax2.grid(True, which="both", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

The charts show that the range of the eigenvalues of the Hessian in the unnormalised data is large, the largest eigenvalue is almost $10^{7}$, and the smallest eigenvalue is almost 0. This indicates that the unnormalised loss function has extremely steep walls and extremely flat valleys: the GD algo would make 0 progress on the floor, and minimal progress on the walls. Conversely, the range of the eigenvalues of the normalised data is comparatively small, this indicates that the normalised loss function is shaped like a quadratic bowl - which is excellent for GD, as it can easily make progress towards the the optimal $\hat{x}$, which minimizes mean squared error. The normalised data exhibits L-smoothness, bounded by $2\lambda_{max}$, and $\mu$ strong convexity, bounded by $2\lambda_{min}$, therefore the optimal step size is $\gamma = \frac{2}{L+\mu}$, or approximately 0.351. This means the error shrinks by a factor of $\frac{\kappa - 1}{\kappa + 1}$, or approximately 0.91 per step.

# Algorithmic Scaling & Complexity: Iterative Gradient Descent vs. Closed-Form Solutions

$x_{k} = x_{k-1} - \gamma \nabla \mathcal{L}(x_{k-1})$

```python
y = train_df_s["Y house price of unit area"].to_numpy()
gamma = 2/(eig_max + eig_min)
```

```python
def loss_function(x,Z,y,n):
    residuals = (Z@x) - y
    return (1 / (2 * n)) * np.dot(residuals, residuals)

def gradient(x):
    return Hessian@x - ((2/331)*(Z.T @ y))
```

```python
import time
x = np.zeros(Z.shape[1])
tol = 1e-6 #error tolerance
max_iters = 10000 

start_time = time.perf_counter()

for i in range(max_iters):
    grad = gradient(x)
    
    #is grad close to 0
    if np.linalg.norm(grad) < tol: 
        print(f"converged in {i} iterations")
        print(f"x is {x}") #check: x is 1 x 6
        break
    x = x - (gamma*grad)

end_time = time.perf_counter()
elapsed_time = end_time - start_time

print(f"Gradient descent execution time: {elapsed_time:.6f} seconds")
```

```python
start_time_normal_eq = time.perf_counter()
x_normal = np.linalg.inv(Z.T @ Z) @ Z.T @ y
end_time_normal = time.perf_counter()
elapsed_time_normal = end_time_normal - start_time_normal_eq

print(f"Solving the normal equations takes {elapsed_time_normal}. Gradient descent is {elapsed_time - elapsed_time_normal} slower in this case.")
print(f"The difference between the analytical solution and the iterative one is {abs(x_normal - x)}")
```

Gradient descent is counter-intuitively slower in this case because the size of the data is relatively small, 331 x 6. Solving the normal equations is $O(d^{3})$ while gradient descent is $O(k\cdot n \cdot d)$. Additionally, np.linalg.inv is highly optimized in Python. Gradient descent offers a superior solution when the size of the data is larger, or when the design matrix is singular/ill-conditioned where exact inversion fails.

```python
import matplotlib.pyplot as plt

# Trackers for history
loss_history = []
grad_norm_history = []

x = np.zeros(Z.shape[1])
gamma = 0.351
tol = 1e-6
max_iters = 10000

for i in range(max_iters):
    grad = gradient(x)
    current_grad_norm = np.linalg.norm(grad)
    
    # Store metrics
    loss_history.append(loss_function(x, Z, y, 331)) # assuming you have your scalar loss function
    grad_norm_history.append(current_grad_norm)
    
    if current_grad_norm < tol:
        print(f"Converged in {i} iterations!")
        break
        
    x = x - (gamma * grad)

# --- Plotting ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Gradient Norm Decay (Log scale is standard for optimization)
ax1.plot(grad_norm_history, color='#1f77b4', lw=2)
ax1.set_yscale('log')
ax1.set_title('Gradient Norm Decay ($||\nabla f(x)||_2$)')
ax1.set_xlabel('Iteration')
ax1.set_ylabel('Norm (Log Scale)')
ax1.grid(True, which="both", ls="--", alpha=0.5)

# Plot 2: Convergence Trajectory
ax2.plot(loss_history, color='#ff7f0e', lw=2)
ax2.set_title('Objective Function Value (MSE)')
ax2.set_xlabel('Iteration')
ax2.set_ylabel('Loss')
ax2.grid(True, ls="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
